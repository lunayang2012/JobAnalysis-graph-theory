## Define US States

### Subtask:
Create a list of all US state abbreviations to use for iteration during job scraping.


**Reasoning**:
The subtask requires creating a Python list of US state abbreviations. I will create a code block to define this list.



```python
us_states = [
    'AL', 'AK', 'AZ', 'AR', 'CA', 'CO', 'CT', 'DE', 'FL', 'GA',
    'HI', 'ID', 'IL', 'IN', 'IA', 'KS', 'KY', 'LA', 'ME', 'MD',
    'MA', 'MI', 'MN', 'MS', 'MO', 'MT', 'NE', 'NV', 'NH', 'NJ',
    'NM', 'NY', 'NC', 'ND', 'OH', 'OK', 'OR', 'PA', 'RI', 'SC',
    'SD', 'TN', 'TX', 'UT', 'VT', 'VA', 'WA', 'WV', 'WI', 'WY'
]

print(f"Created a list of {len(us_states)} US state abbreviations.")
```

## Initialize Master DataFrame

### Subtask:
Create an empty pandas DataFrame that will be used to store job listings from all states.


**Reasoning**:
To store job listings from all states, I need to initialize an empty pandas DataFrame. This will involve importing the pandas library and then creating the DataFrame.



```python
import pandas as pd

all_us_jobs_df = pd.DataFrame()

print("Empty DataFrame 'all_us_jobs_df' created successfully.")
```

## Scrape Jobs for Each State

### Subtask:
Iterate through the list of US states. For each state, call the `scrape_jobs` function with the `LOCATION` parameter set to the current state. Append the results to the master DataFrame.


**Reasoning**:
To perform the job scraping, I need to implement a loop that iterates through each US state, calls the `scrape_jobs` function for that state, and appends the results to the master DataFrame. This step addresses all the instructions provided.



```python
from jobspy import scrape_jobs

# Loop through each state and scrape jobs
for state in us_states:
    print(f"\n--- Scraping jobs for {state} ---")
    try:
        state_jobs_df = scrape_jobs(
            site_name=SITES,
            search_term=SEARCH_TERM,
            location=state,
            results_wanted=RESULTS_WANTED,
            hours_old=HOURS_OLD,
            distance=DISTANCE,
            country_indeed='USA'
        )

        if not state_jobs_df.empty:
            initial_rows = len(all_us_jobs_df)
            all_us_jobs_df = pd.concat([all_us_jobs_df, state_jobs_df], ignore_index=True)
            added_rows = len(all_us_jobs_df) - initial_rows
            print(f"Found {len(state_jobs_df)} jobs in {state}. Added {added_rows} jobs to the master list.")
        else:
            print(f"No jobs found for '{SEARCH_TERM}' in {state}.")

        # Remove duplicates based on 'id' if 'id' column exists
        if 'id' in all_us_jobs_df.columns:
            all_us_jobs_df.drop_duplicates(subset=['id'], inplace=True)

        print(f"Total unique jobs collected so far: {len(all_us_jobs_df)}")

    except Exception as e:
        print(f"Error scraping jobs for {state}: {e}")

print("\n--- Scraping complete ---")
print(f"Finished processing all US states. Total unique jobs collected: {len(all_us_jobs_df)}")
```

## Scrape Jobs for All Search Terms and States

### Subtask:
Iterate through the `search_terms` list. For each search term, iterate through the `us_states` list. Call the `scrape_jobs` function with the current `search_term` and `LOCATION` (state), and append the results to the `all_search_terms_jobs_df` DataFrame.

**Reasoning**:
To execute the job scraping for all search terms and states, I need to implement nested loops: one for `search_terms` and another for `us_states`. Inside the inner loop, I will call `scrape_jobs` with the respective parameters and concatenate the results to the master DataFrame, `all_search_terms_jobs_df`, while handling duplicates.



```python
from jobspy import scrape_jobs

# Loop through each search term
for term in search_terms:
    print(f"\n=== Starting scraping for search term: '{term}' ===")
    # Loop through each state and scrape jobs for the current search term
    for state in us_states:
        print(f"\n--- Scraping jobs for '{term}' in {state} ---")
        try:
            state_term_jobs_df = scrape_jobs(
                site_name=SITES,
                search_term=term,
                location=state,
                results_wanted=RESULTS_WANTED,
                hours_old=HOURS_OLD,
                distance=DISTANCE,
                country_indeed='USA'
            )

            if not state_term_jobs_df.empty:
                initial_rows = len(all_search_terms_jobs_df)
                all_search_terms_jobs_df = pd.concat([all_search_terms_jobs_df, state_term_jobs_df], ignore_index=True)

                # Remove duplicates based on 'id' if 'id' column exists
                if 'id' in all_search_terms_jobs_df.columns:
                    all_search_terms_jobs_df.drop_duplicates(subset=['id'], inplace=True)

                added_rows = len(all_search_terms_jobs_df) - initial_rows
                print(f"Found {len(state_term_jobs_df)} jobs for '{term}' in {state}. Added {added_rows} unique jobs to the master list.")
            else:
                print(f"No jobs found for '{term}' in {state}.")

            print(f"Total unique jobs collected so far for all terms: {len(all_search_terms_jobs_df)}")

        except Exception as e:
            print(f"Error scraping jobs for '{term}' in {state}: {e}")

print("\n--- Overall Scraping complete ---")
print(f"Finished processing all US states for all search terms. Total unique jobs collected: {len(all_search_terms_jobs_df)}")
```

## Split and Save Job Data per Search Term

### Subtask:
For each search term, filter the combined job listings DataFrame, split the filtered data into three approximately equal parts, and save each part to a separate Excel file, then download these files.


**Reasoning**:
To address the subtask, I need to iterate through each search term, filter the main DataFrame, split the filtered data into three parts, save each part to an Excel file with a specific naming convention, and then download these files. I will also re-include the `clean_xml_string` function to prevent `IllegalCharacterError` during Excel saving.



```python
import re
from google.colab import files
import numpy as np

# Function to clean strings by removing characters not allowed in XML 1.0
def clean_xml_string(text):
    if isinstance(text, str):
        # This regex keeps only printable ASCII characters (0x20-0x7E)
        # and the allowed XML whitespace characters (tab, newline, carriage return).
        # It removes all other characters, including C0/C1 control characters and non-ASCII characters.
        cleaned_text = re.sub(r'[\x09\x0A\x0D\x20-\x7E]', '', text)
        return cleaned_text
    return text

# Loop through each search term
for term in search_terms:
    print(f"\n--- Processing search term: '{term}' ---")

    # Filter the combined DataFrame for the current search term
    filtered_df = all_search_terms_jobs_df[all_search_terms_jobs_df['title'].str.contains(term, case=False, na=False) |
                                           all_search_terms_jobs_df['description'].str.contains(term, case=False, na=False)].copy()

    # Apply cleaning to the filtered DataFrame before saving
    for col in filtered_df.select_dtypes(include=['object']).columns:
        filtered_df[col] = filtered_df[col].apply(clean_xml_string)

    if filtered_df.empty:
        print(f"No jobs found for '{term}' after filtering. Skipping.")
        continue

    total_rows = len(filtered_df)
    rows_per_part = total_rows // 3
    remainder = total_rows % 3

    # Calculate split points
    split_point1 = rows_per_part + (1 if remainder > 0 else 0)
    split_point2 = split_point1 + rows_per_part + (1 if remainder > 1 else 0)

    # Split the DataFrame into three parts
    part1 = filtered_df.iloc[:split_point1]
    part2 = filtered_df.iloc[split_point1:split_point2]
    part3 = filtered_df.iloc[split_point2:]

    # Define base filename without spaces
    base_term_filename = term.replace(' ', '_')

    # Save and download each part
    parts_to_process = [(part1, 1), (part2, 2), (part3, 3)]
    for df_part, part_num in parts_to_process:
        if not df_part.empty:
            file_name = f'all_us_{base_term_filename}_jobs_part{part_num}.xlsx'
            try:
                df_part.to_excel(file_name, index=False)
                print(f"Saved '{file_name}' with {len(df_part)} rows.")
                files.download(file_name)
                print(f"Initiated download for '{file_name}'.")
            except Exception as e:
                print(f"Error saving or downloading '{file_name}': {e}")
        else:
            print(f"Part {part_num} for '{term}' is empty. Skipping save and download.")

print("\n--- All search terms processed ---")
```
